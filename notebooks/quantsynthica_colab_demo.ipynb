## 📊 Section 1: Deep Corporate Fundamentals

```python
# Cell 2: Key Financial & Valuation Ratios
import quantsynthica as qs
import pandas as pd

ratios = qs.get_ratios("RELIANCE")

print(f"=== Fundamental Ratios for {ratios.symbol} ===")
print(f"Trailing P/E:       {ratios.pe_trailing:.2f}")
print(f"Forward P/E:        {ratios.pe_forward:.2f}")
print(f"Price / Book:       {ratios.price_to_book:.2f}")
print(f"EV / EBITDA:        {ratios.ev_to_ebitda:.2f}")
print(f"EV / Revenue:       {ratios.ev_to_revenue:.2f}")
print(f"Operating Margin:   {ratios.operating_margin_pct}%")
print(f"Net Profit Margin:  {ratios.net_margin_pct}%")
print(f"Debt / Equity:      {ratios.debt_to_equity}")
print(f"Revenue Growth:     {ratios.revenue_growth_pct}%")
print(f"Earnings Growth:    {ratios.earnings_growth_pct}%")
```

```python
# Cell 4: Balance Sheet (Assets, Liabilities & Equity)
balance_sheet = qs.get_balance_sheet("INFY")

key_bs_rows = [r for r in ["Total Assets", "Total Liabilities Net Minority Interest", "Stockholders Equity", "Working Capital", "Total Debt", "Cash And Cash Equivalents"] if r in balance_sheet.index]
balance_sheet.loc[key_bs_rows]
```

```python
# Cell 5: Cash Flow Statement & Free Cash Flow (FCF)
cashflow = qs.get_cashflow("RELIANCE")

key_cf_rows = [r for r in ["Operating Cash Flow", "Capital Expenditure", "Free Cash Flow", "Financing Cash Flow", "Investing Cash Flow"] if r in cashflow.index]
cashflow.loc[key_cf_rows]
```

```python
# Cell 6: Wall Street & Dalal Street Consensus Price Targets
targets = qs.get_analyst_targets("RELIANCE")

print(f"=== Analyst Consensus for {targets.symbol} ===")
print(f"Current Price:      ₹{targets.current_price}")
print(f"Target Mean:        ₹{targets.target_mean_price} ({(targets.target_mean_price/targets.current_price - 1)*100:.1f}% potential)")
print(f"Target High:        ₹{targets.target_high_price}")
print(f"Target Low:         ₹{targets.target_low_price}")
print(f"Consensus Rating:   {targets.recommendation.upper()}")
print(f"Analyst Coverage:   {targets.number_of_analysts} institutional analysts")
```

```python
# Cell 7: Company Profile & Business Overview
profile = qs.get_profile("HDFCBANK")

print(f"Company:     {profile.name}")
print(f"Sector:      {profile.sector} | Industry: {profile.industry}")
print(f"Employees:   {profile.employees:,}")
print(f"Website:     {profile.website}\n")
print("Business Summary:")
print(profile.summary[:500] + "...")
```

```python
# Cell 8: Historical Dividend Payments
import matplotlib.pyplot as plt

stock = qs.Ticker("TCS")
divs = stock.dividends()

if not divs.empty:
    plt.figure(figsize=(10, 4))
    divs.tail(15).plot(kind="bar", color="#10b981")
    plt.title("TCS Recent Dividend Payouts (INR)")
    plt.xlabel("Date")
    plt.ylabel("Dividend / Share")
    plt.xticks(rotation=45)
    plt.grid(True, alpha=0.3)
    plt.show()
else:
    print("No dividend history recorded.")
```

## 🏛️ Section 2: 4-in-1 Valuation Models (DCF, Piotroski, Altman Z, Graham)

```python
# Cell 9: 4-in-1 Comprehensive Valuation Suite
val = qs.get_valuation("RELIANCE")

print("=== 4-in-1 Institutional Valuation ===")
print(f"DCF Fair Value:     ₹{val.dcf.fair_value_per_share} (Upside: {val.dcf.upside_pct}%)")
print(f"Piotroski F-Score:  {val.piotroski.f_score} / 9 ({val.piotroski.rating})")
print(f"Altman Z-Score:     {val.altman_z.z_score} ({val.altman_z.zone})")
print(f"Graham Number:      ₹{val.graham.graham_number} (Margin of Safety: {val.graham.margin_of_safety_pct}%)")
```

```python
# Cell 10: Custom Discounted Cash Flow (DCF) Sensitivity Modeling
reliance = qs.Ticker("RELIANCE")

print("DCF Fair Value with different growth rates:")
for g in [0.08, 0.10, 0.12, 0.14, 0.16]:
    d = reliance.dcf(growth_rate=g, discount_rate=0.105, terminal_growth_rate=0.035)
    print(f"  Growth {int(g*100):>2}%: Fair Value ₹{d.fair_value_per_share:>7.2f} (Upside: {d.upside_pct:>5.1f}%)")
```

## 🔬 Section 3: Technicals, Screener & Portfolio Optimization

```python
# Cell 11: Vectorized Technical Indicators
tech = qs.get_technicals("INFY", period="6mo")

print(f"INFY Last Close:    ₹{tech.latest_close}")
print(f"RSI (14):           {tech.rsi_14:.2f}")
print(f"SMA 20 vs SMA 50:   ₹{tech.sma_20:.2f} vs ₹{tech.sma_50:.2f}")
print(f"MACD Line / Signal: {tech.macd.macd_line:.2f} / {tech.macd.signal_line:.2f}")
print(f"Bollinger Bands:    [₹{tech.bollinger_bands.lower:.2f} — ₹{tech.bollinger_bands.upper:.2f}]")
```

```python
# Cell 12: TradingView Multi-Asset Screener
results = qs.screener.value_stocks(market="india", limit=10)
df_screener = pd.DataFrame(results.data)
df_screener[["ticker", "name", "close", "price_earnings_ttm", "market_cap_basic"]].head(10)
```

```python
# Cell 13: Markowitz Efficient Frontier Portfolio Optimization
portfolio = ["RELIANCE", "TCS", "INFY", "HDFCBANK", "ICICIBANK"]
opt = qs.optimize_portfolio(portfolio, objective="max_sharpe", period="1y")

print("=== Optimal Portfolio Allocations (Max Sharpe) ===")
print(f"Expected Annual Return: {opt.optimal_portfolio.expected_annual_return_pct}%")
print(f"Annual Volatility:      {opt.optimal_portfolio.annual_volatility_pct}%")
print(f"Sharpe Ratio:           {opt.optimal_portfolio.sharpe_ratio:.2f}\n")

for alloc in opt.allocations:
    print(f"  {alloc.symbol:<12}: {alloc.weight_pct:>6.2f}%")
```

```python
# Cell 14: Algorithmic Backtesting (SMA Crossover Strategy)
bt = qs.backtest(
    "AAPL",
    strategy="sma_crossover",
    period="2y",
    params={"fast_period": 10, "slow_period": 30},
    initial_capital=100000.0,
)

print("=== Strategy Backtest Performance ===")
print(f"Initial Capital:  ${bt.summary.initial_capital:,.2f}")
print(f"Final Equity:     ${bt.summary.final_equity:,.2f}")
print(f"Total Return:     {bt.summary.total_return_pct}%")
print(f"Total Trades:     {bt.summary.total_trades}")
print(f"Win Rate:         {bt.summary.win_rate_pct}%")
```
